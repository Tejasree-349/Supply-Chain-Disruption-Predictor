```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Add project root to Python's import path
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_loader import load_data

sns.set_theme(style="whitegrid")

print("Python version:", sys.version)
print("Project root:", PROJECT_ROOT)
print("Libraries imported successfully!")
```

```python
df = load_data()

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
print("DATASET INFORMATION")
print("=" * 50)

df.info()

print("\nNUMERICAL SUMMARY")
display(df.describe().T)

print("\nCOLUMN NAMES")
for column in df.columns:
    print(column)
missing = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(df) * 100).round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing_Percentage",
    ascending=False
)

display(missing_summary[missing_summary["Missing_Count"] > 0])

print("Duplicate rows:", df.duplicated().sum())
```

```python
target = "Late_delivery_risk"

if target not in df.columns:
    raise ValueError(
        f"Target column '{target}' not found. "
        f"Available columns: {df.columns.tolist()}"
    )

target_counts = df[target].value_counts().sort_index()

print("Target counts:")
display(target_counts.rename("Number_of_Orders").to_frame())

print("\nTarget percentages:")
display(
    (df[target].value_counts(normalize=True).sort_index() * 100)
    .round(2)
    .rename("Percentage")
    .to_frame()
)

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x=target
)

plt.title("Distribution of Late Delivery Risk")
plt.xlabel("Late Delivery Risk")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

```python
possible_leakage_columns = [
    column for column in df.columns
    if any(
        keyword in column.lower()
        for keyword in [
            "delivery_status",
            "days_for_shipping_(real)",
            "actual_delivery",
            "date_of_delivery"
        ]
    )
]

print("Potential data leakage columns:")

for column in possible_leakage_columns:
    print("-", column)
```
